```python

import pandas as pd

# =========================
# Part 1: XPT -> CSV
# =========================

bp_df = pd.read_sas("BPXO_L (1).xpt", format="xport")
bp_df.to_csv("BPXO_L (1).csv", index=False)

bmx_df = pd.read_sas("BMX_L (1).xpt", format="xport")
bmx_df.to_csv("BMX_L (1).csv", index=False)

print("Conversion completed!")
print("BPXO Shape:", bp_df.shape)
print("BMX Shape:", bmx_df.shape)


# =========================
# Part 2: Read CSV
# =========================

bp_df = pd.read_csv("BPXO_L (1).csv")
bmx_df = pd.read_csv("BMX_L (1).csv")


# =========================
# Part 3: Target variables
# =========================

bp_targets = [
    "BPXOSY1",
    "BPXOSY2",
    "BPXOSY3",
    "BPXODI1",
    "BPXODI2",
    "BPXODI3"
]

bmx_targets = ["BMXBMI"]


# =========================
# Part 4: Extract SEQN
# =========================

result = pd.DataFrame()

counts = {}

# Process BPXO variables
for target in bp_targets:

    filtered_df = bp_df[bp_df[target].notna()]

    seqn = filtered_df["SEQN"].reset_index(drop=True)

    result[target] = seqn

    counts[target] = len(seqn)


# Process BMX variables
for target in bmx_targets:

    filtered_df = bmx_df[bmx_df[target].notna()]

    seqn = filtered_df["SEQN"].reset_index(drop=True)

    result[target] = seqn

    counts[target] = len(seqn)


# =========================
# Part 5: Save output
# =========================

result.to_csv("Filtered_Column1.csv", index=False)


# =========================
# Part 6: Print counts
# =========================

print("\n=== Valid observations ===")

for target, count in counts.items():
    print(f"{target}: {count}")

print("\nFiltered_Column1.csv created!")
```

```python

import pandas as pd
import numpy as np

# ============================================================
# 1. Load data
# ============================================================

df = pd.read_csv("SMQ_L (1).csv")

# Keep only required variables
df = df[["SEQN", "SMQ020", "SMQ040"]].copy()

# Convert variables to numeric
for col in ["SMQ020", "SMQ040"]:
    df[col] = pd.to_numeric(df[col], errors="coerce")

# ============================================================
# 2. Create Smoking Status
# ============================================================

conditions = [
    df["SMQ020"] == 2,

    (df["SMQ020"] == 1) &
    (df["SMQ040"] == 3),

    (df["SMQ020"] == 1) &
    (df["SMQ040"].isin([1, 2]))
]

groups = [
    "Never Smoker",
    "Former Smoker",
    "Current Smoker"
]

df["Smoking_Status"] = np.select(
    conditions,
    groups,
    default="Missing"
)

# ============================================================
# 3. One-hot encoding
# ============================================================

smoking_groups = [
    "Never Smoker",
    "Former Smoker",
    "Current Smoker"
]

for group in smoking_groups:

    df[group] = (
        df["Smoking_Status"] == group
    ).astype(int)

# ============================================================
# 4. Quality checks
# ============================================================

print("\n=== Smoking Status Counts ===")

print(
    df["Smoking_Status"]
    .value_counts(dropna=False)
)

print("\n=== One-Hot Matrix Check ===")

print(
    df[smoking_groups]
    .sum(axis=1)
    .value_counts()
)

print("\n=== Total Participants ===")

print(len(df))

# ============================================================
# 5. Export
# ============================================================

df.to_csv(
    "SMQ_STATUS.csv",
    index=False
)

print("\nSMQ_STATUS.csv successfully created!")
```

```python

import pandas as pd
import numpy as np

# ==========================================
# 1. Load data
# ==========================================

df = pd.read_csv("PAQ_L (1).csv")

# ==========================================
# 2. Clean frequency variables
# ==========================================

frequency_columns = ["PAD790Q", "PAD810Q"]

for col in frequency_columns:

    df[col] = pd.to_numeric(df[col], errors="coerce")

    # Convert tiny numerical artifacts to zero
    df.loc[df[col].abs() < 1e-50, col] = 0

    # Special missing codes
    df.loc[df[col].isin([7777, 9999]), col] = np.nan


# ==========================================
# 3. Clean duration variables
# ==========================================

duration_columns = ["PAD800", "PAD820"]

for col in duration_columns:

    df[col] = pd.to_numeric(df[col], errors="coerce")

    # Special missing codes
    df.loc[df[col].isin([7777, 9999]), col] = np.nan


# ==========================================
# 4. Clean frequency units
# ==========================================

def clean_unit(x):

    if pd.isna(x):
        return np.nan

    if isinstance(x, bytes):
        x = x.decode()

    x = str(x).strip().upper()

    # Handle strings such as b'W'
    if x.startswith("B'") and x.endswith("'"):
        x = x[2:-1]

    if x.startswith('B"') and x.endswith('"'):
        x = x[2:-1]

    x = x.strip()

    return x if x in ["W", "D", "M", "Y"] else np.nan


df["PAD790U"] = df["PAD790U"].apply(clean_unit)
df["PAD810U"] = df["PAD810U"].apply(clean_unit)


# ==========================================
# 5. Convert frequency to times per week
# ==========================================

unit_factor = {
    "W": 1,
    "D": 7,
    "M": 12 / 52,
    "Y": 1 / 52
}

def frequency_per_week(freq, unit):

    # Missing frequency
    if pd.isna(freq):
        return np.nan

    # Zero frequency is valid
    if freq == 0:
        return 0.0

    # Positive frequency requires a valid unit
    if pd.isna(unit):
        return np.nan

    return freq * unit_factor[unit]


df["Moderate_Freq_Week"] = [
    frequency_per_week(f, u)
    for f, u in zip(df["PAD790Q"], df["PAD790U"])
]

df["Vigorous_Freq_Week"] = [
    frequency_per_week(f, u)
    for f, u in zip(df["PAD810Q"], df["PAD810U"])
]


# ==========================================
# 6. Calculate weekly minutes
# ==========================================

df["Moderate_Min_Week"] = (
    df["Moderate_Freq_Week"] * df["PAD800"]
)

df["Vigorous_Min_Week"] = (
    df["Vigorous_Freq_Week"] * df["PAD820"]
)

# Zero frequency means zero weekly minutes,
# even if duration is missing.

df.loc[df["Moderate_Freq_Week"] == 0,
       "Moderate_Min_Week"] = 0

df.loc[df["Vigorous_Freq_Week"] == 0,
       "Vigorous_Min_Week"] = 0


# ==========================================
# 7. Calculate MET-minutes per week
# ==========================================

df["Moderate_MET"] = 4 * df["Moderate_Min_Week"]

df["Vigorous_MET"] = 8 * df["Vigorous_Min_Week"]


# ==========================================
# 8. Calculate Total MET
# ==========================================

# If both components are valid:
#     Total = Moderate + Vigorous
#
# If only Moderate is valid:
#     Total = Moderate
#
# If only Vigorous is valid:
#     Total = Vigorous
#
# If both are missing:
#     Total = Missing
#
# Zero values are preserved.

df["Total_MET"] = df[
    ["Moderate_MET", "Vigorous_MET"]
].sum(axis=1, min_count=1)


# ==========================================
# 9. Remove participants with both MET missing
# ==========================================

before = len(df)

df = df.dropna(
    subset=["Moderate_MET", "Vigorous_MET"],
    how="all"
).copy()

removed = before - len(df)


# ==========================================
# 10. Export final dataset
# ==========================================

output_columns = [
    "SEQN",
    "Moderate_MET",
    "Vigorous_MET",
    "Total_MET"
]

df[output_columns].to_csv(
    "PAQ_MET.csv",
    index=False
)


# ==========================================
# 11. Quality check
# ==========================================

print("========== DATA QUALITY CHECK ==========")

print(f"Original participants : {before}")
print(f"Removed participants : {removed}")
print(f"Final participants  : {len(df)}")

print("\nMissing values:")
print(df[output_columns].isna().sum())

print("\nZero MET values:")
print("Moderate MET:", (df["Moderate_MET"] == 0).sum())
print("Vigorous MET:", (df["Vigorous_MET"] == 0).sum())
print("Total MET:", (df["Total_MET"] == 0).sum())

print("\nSaved successfully: PAQ_MET.csv")
```

```python

import pandas as pd
import numpy as np

# ==========================================
# 1. Load PAQ_MET.csv
# ==========================================

df = pd.read_csv("PAQ_MET.csv")


# ==========================================
# 2. Define physical activity groups
# ==========================================

df["Inactive"] = (df["Total_MET"] == 0).astype(int)

df["Low"] = (
    (df["Total_MET"] > 0) &
    (df["Total_MET"] < 600)
).astype(int)

df["Active"] = (
    (df["Total_MET"] >= 600) &
    (df["Total_MET"] < 1500)
).astype(int)

df["Highly_Active"] = (
    df["Total_MET"] >= 1500
).astype(int)


# ==========================================
# 3. Save updated PAQ_MET.csv
# ==========================================

df.to_csv("PAQ_MET.csv", index=False)


# ==========================================
# 4. Quality check
# ==========================================

group_columns = [
    "Inactive",
    "Low",
    "Active",
    "Highly_Active"
]

print("========== GROUP COUNTS ==========")

for col in group_columns:
    print(f"{col:15s}: {df[col].sum():5d}")

print("\n========== CHECK ==========")

print("Total participants:", len(df))

print(
    "Participants assigned to a group:",
    df[group_columns].sum(axis=1).eq(1).sum()
)

print(
    "Participants with no group:",
    df[group_columns].sum(axis=1).eq(0).sum()
)

print(
    "Participants with multiple groups:",
    df[group_columns].sum(axis=1).gt(1).sum()
)

print("\nUpdated PAQ_MET.csv successfully!")
```

```python

import pandas as pd
import matplotlib.pyplot as plt

# Read CSV file
smq = pd.read_csv("SMQ_STATUS.csv")

# Smoking groups
smoking_groups = [
    "Never Smoker",
    "Former Smoker",
    "Current Smoker"
]

# Frequency
smoking_counts = smq[smoking_groups].sum()

# Percentage
total = smoking_counts.sum()
smoking_percent = smoking_counts / total * 100

# Frequency table
smoking_table = pd.DataFrame({
    "Frequency": smoking_counts,
    "Percentage (%)": smoking_percent
})

print("\nSmoking Status Frequency Distribution:")
print(smoking_table.round(2))

# Plot
fig, ax = plt.subplots(figsize=(8, 5))

bars = ax.bar(
    smoking_counts.index,
    smoking_counts.values,
    color="steelblue"
)

# Add percentage labels
for bar, percentage in zip(bars, smoking_percent):
    ax.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f"{percentage:.1f}%",
        ha="center",
        va="bottom",
        fontsize=11
    )

ax.set_title("Frequency Distribution of Smoking Status")
ax.set_xlabel("Smoking Status")
ax.set_ylabel("Number of Participants")

plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd

files = {
    "BMX_L (1).csv": ["BMXBMI"],
    "BP_average.csv": ["SYS_AVG", "DIA_AVG"],
    "GHB_L (1).csv": ["LBXGH"],
    "GLU_L (1).csv": ["LBXGLU"],
    "HDL_L (1).csv": ["LBDHDD"],
    "TCHOL_L (1).csv": ["LBXTC"],
    "TRIGLY_L (1).csv": ["LBXTLG"],
}

results = []

for filename, variables in files.items():

    df = pd.read_csv(filename)

    for var in variables:

        x = pd.to_numeric(df[var], errors="coerce").dropna()

        Q1 = x.quantile(0.25)
        Q3 = x.quantile(0.75)

        results.append({
            "Variable": var,
            "N": x.count(),
            "Missing": df[var].isna().sum(),
            "Mean": x.mean(),
            "Median": x.median(),
            "SD": x.std(),
            "Q1": Q1,
            "Q3": Q3,
            "IQR": Q3 - Q1,
            "Minimum": x.min(),
            "Maximum": x.max(),
            "Range": x.max() - x.min()
        })

summary = pd.DataFrame(results)

display(summary.round(2))

summary.to_csv("Response_Variables_Center_Spread.csv", index=False)

print("Summary table saved successfully.")
```

```python

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ==================================================
# 1. Load datasets
# ==================================================

bmx = pd.read_csv("BMX_L (1).csv")
ghb = pd.read_csv("GHB_L (1).csv")
smq = pd.read_csv("SMQ_STATUS.csv")
demo = pd.read_csv("DEMO_L.csv")

# ==================================================
# 2. Merge datasets using SEQN
# ==================================================

df = bmx[["SEQN", "BMXBMI"]].merge(
    ghb[["SEQN", "LBXGH"]],
    on="SEQN",
    how="outer"
)

df = df.merge(
    smq[["SEQN", "Smoking_Status"]],
    on="SEQN",
    how="left"
)

df = df.merge(
    demo[["SEQN", "RIDAGEYR"]],
    on="SEQN",
    how="left"
)

# ==================================================
# 3. Restrict to adults (Age >= 18)
# ==================================================

df = df[df["RIDAGEYR"] >= 18].copy()

# ==================================================
# 4. Define smoking categories
# ==================================================

smoking_order = [
    "Never Smoker",
    "Former Smoker",
    "Current Smoker"
]

# ==================================================
# 5. Output directory
# ==================================================

output_dir = Path("Smoking_Comparison")
output_dir.mkdir(exist_ok=True)

# ==================================================
# 6. Boxplot function
# ==================================================

def create_boxplot(data, response, ylabel, title, filename):

    # Select relevant variables
    plot_data = data[["Smoking_Status", response]].copy()

    # Remove missing observations for this plot
    plot_data = plot_data.dropna()

    # Keep valid smoking categories
    plot_data = plot_data[
        plot_data["Smoking_Status"].isin(smoking_order)
    ]

    # Extract data for each smoking group
    grouped_data = [
        plot_data.loc[
            plot_data["Smoking_Status"] == category,
            response
        ].values
        for category in smoking_order
    ]

    # Sample size per group
    counts = plot_data["Smoking_Status"].value_counts()

    labels = [
        f"{category}\n(n={counts.get(category, 0)})"
        for category in smoking_order
    ]

    # Create figure
    fig, ax = plt.subplots(figsize=(10, 6))

    ax.boxplot(
    grouped_data,
    tick_labels=labels,
    showfliers=True
    )

    # Titles and axes
    ax.set_title(title, fontsize=14, pad=15)
    ax.set_xlabel("Smoking Status", fontsize=12)
    ax.set_ylabel(ylabel, fontsize=12)

    # Grid
    ax.grid(
        axis="y",
        linestyle="--",
        alpha=0.4
    )

    # Layout
    plt.tight_layout()

    # Save figure
    plt.savefig(
        output_dir / filename,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    # Print sample sizes
    print(f"\n{title}")
    print(f"Total observations: {len(plot_data)}")
    print(counts.reindex(smoking_order, fill_value=0))


# ==================================================
# 7. Figure 1: BMI by Smoking Status
# ==================================================

create_boxplot(
    data=df,
    response="BMXBMI",
    ylabel="Body Mass Index (kg/m²)",
    title="Distribution of Body Mass Index by Smoking Status",
    filename="Figure_1_BMI_by_Smoking.png"
)

# ==================================================
# 8. Figure 2: HbA1c by Smoking Status
# ==================================================

create_boxplot(
    data=df,
    response="LBXGH",
    ylabel="Glycohemoglobin (HbA1c, %)",
    title="Distribution of Glycohemoglobin by Smoking Status",
    filename="Figure_2_HbA1c_by_Smoking.png"
)

# ==================================================
# 9. Completion message
# ==================================================

print("\nBoth figures have been generated successfully.")
print(f"Figures saved in: {output_dir.resolve()}")
```

```python

import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path

# ==================================================
# 1. Load datasets
# ==================================================

bmx = pd.read_csv("BMX_L (1).csv")
ghb = pd.read_csv("GHB_L (1).csv")
paq = pd.read_csv("PAQ_MET.csv")
demo = pd.read_csv("DEMO_L.csv")

# ==================================================
# 2. Merge datasets using SEQN
# ==================================================

df = bmx[["SEQN", "BMXBMI"]].merge(
    ghb[["SEQN", "LBXGH"]],
    on="SEQN",
    how="outer"
)

df = df.merge(
    paq[["SEQN", "Physical_Activity_Group"]],
    on="SEQN",
    how="left"
)

df = df.merge(
    demo[["SEQN", "RIDAGEYR"]],
    on="SEQN",
    how="left"
)

# ==================================================
# 3. Restrict to adults (Age >= 18)
# ==================================================

df = df[df["RIDAGEYR"] >= 18].copy()

# ==================================================
# 4. Define physical activity categories
# ==================================================

activity_order = [
    "Inactive",
    "Low",
    "Active",
    "Highly Active"
]

# ==================================================
# 5. Output directory
# ==================================================

output_dir = Path("Physical_Activity_Comparison")
output_dir.mkdir(exist_ok=True)

# ==================================================
# 6. Boxplot function
# ==================================================

def create_boxplot(data, response, ylabel, title, filename):

    plot_data = data[
        ["Physical_Activity_Group", response]
    ].dropna()

    plot_data = plot_data[
        plot_data["Physical_Activity_Group"].isin(activity_order)
    ]

    grouped_data = [
        plot_data.loc[
            plot_data["Physical_Activity_Group"] == category,
            response
        ].values
        for category in activity_order
    ]

    counts = plot_data["Physical_Activity_Group"].value_counts()

    labels = [
        f"{category}\n(n={counts.get(category, 0)})"
        for category in activity_order
    ]

    # Create figure
    fig, ax = plt.subplots(figsize=(11, 6))

    ax.boxplot(
        grouped_data,
        tick_labels=labels,
        showfliers=True,
        widths=0.6
    )

    ax.set_title(title, fontsize=14, pad=15)
    ax.set_xlabel("Physical Activity Level", fontsize=12)
    ax.set_ylabel(ylabel, fontsize=12)

    ax.grid(axis="y", linestyle="--", alpha=0.4)

    plt.tight_layout()

    plt.savefig(
        output_dir / filename,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()

    # Descriptive statistics
    summary = (
        plot_data.groupby("Physical_Activity_Group")[response]
        .agg(
            N="count",
            Mean="mean",
            Median="median",
            SD="std",
            Q1=lambda x: x.quantile(0.25),
            Q3=lambda x: x.quantile(0.75),
            Minimum="min",
            Maximum="max"
        )
        .reindex(activity_order)
    )

    summary["IQR"] = summary["Q3"] - summary["Q1"]

    print(f"\nDescriptive statistics: {response}")
    print(summary.round(3))

    summary.to_csv(
        output_dir / f"{response}_Physical_Activity_Statistics.csv"
    )

    return summary

# ==================================================
# 7. Figure 1: BMI by Physical Activity
# ==================================================

BMI_PA_summary = create_boxplot(
    data=df,
    response="BMXBMI",
    ylabel="Body Mass Index (kg/m²)",
    title="Distribution of Body Mass Index by Physical Activity Level",
    filename="Figure_1_BMI_by_Physical_Activity.png"
)

# ==================================================
# 8. Figure 2: HbA1c by Physical Activity
# ==================================================

HbA1c_PA_summary = create_boxplot(
    data=df,
    response="LBXGH",
    ylabel="Glycohemoglobin (HbA1c, %)",
    title="Distribution of Glycohemoglobin by Physical Activity Level",
    filename="Figure_2_HbA1c_by_Physical_Activity.png"
)

# ==================================================
# 9. Completion
# ==================================================

print("\nBoth physical activity figures were generated.")
print(f"Output directory: {output_dir.resolve()}")
```

```python

import pandas as pd

# Load existing physical activity matrix
paq = pd.read_csv("PAQ_MET.csv")

# One-hot encoded columns
activity_columns = [
    "Inactive",
    "Low",
    "Active",
    "Highly_Active"
]

# Convert one-hot matrix into categorical variable
paq["Physical_Activity_Group"] = (
    paq[activity_columns]
    .idxmax(axis=1)
)

# Participants with no active category (all zeros) -> Missing
paq.loc[
    paq[activity_columns].sum(axis=1) == 0,
    "Physical_Activity_Group"
] = pd.NA

# Save updated file
paq.to_csv("PAQ_MET.csv", index=False)

# Check results
print("\nPhysical Activity Group Counts:")
print(paq["Physical_Activity_Group"].value_counts(dropna=False))

print("\nUpdated PAQ_MET.csv successfully!")
```

```python

import pandas as pd
import numpy as np

# 1. Load data
demo = pd.read_csv("DEMO_L.csv")
bmx = pd.read_csv("BMX_L (1).csv")

# 2. Merge using SEQN
df = demo[["SEQN", "RIDAGEYR", "WTMEC2YR"]].merge(
    bmx[["SEQN", "BMXBMI"]],
    on="SEQN",
    how="inner",
    validate="one_to_one"
)

# 3. Restrict to adults (age >= 18)
df = df[df["RIDAGEYR"] >= 18].copy()

# 4. Keep valid BMI and weight observations
df["BMXBMI"] = pd.to_numeric(df["BMXBMI"], errors="coerce")
df["WTMEC2YR"] = pd.to_numeric(df["WTMEC2YR"], errors="coerce")

df = df[
    df["BMXBMI"].notna() &
    df["WTMEC2YR"].notna() &
    (df["WTMEC2YR"] > 0)
].copy()

# 5. Calculate summaries
n = len(df)

unweighted_mean = df["BMXBMI"].mean()

weighted_mean = np.average(
    df["BMXBMI"],
    weights=df["WTMEC2YR"]
)

# 6. Difference
difference = weighted_mean - unweighted_mean

# 7. Print results
print("BMI Summary (Adults >= 18)")
print("-" * 40)
print(f"Sample size:       {n}")
print(f"Unweighted mean:   {unweighted_mean:.3f}")
print(f"Weighted mean:     {weighted_mean:.3f}")
print(f"Difference:         {difference:.3f}")

```
